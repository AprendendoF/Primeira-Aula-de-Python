```python
salario_atual = float(input("Informe o salário do colaborador: R$ "))
salario_original = salario_atual

if salario_atual <= 280.00:
    percentual_aumento = 20
elif salario_atual <= 700.00:
    percentual_aumento = 15
elif salario_atual <= 1500.00:
    percentual_aumento = 10
else:
    percentual_aumento = 5

aumento = salario_atual * (percentual_aumento / 100)
novo_salario = salario_atual + aumento

print(f"Salário antes do reajuste: R$ {salario_original:.2f}")
print(f"Percentual de aumento aplicado: {percentual_aumento}%")
print(f"Valor do aumento: R$ {aumento:.2f}")
print(f"Novo salário, após o aumento: R$ {novo_salario:.2f}")
```

```python
C = "Celsius"
F = "Fahrenheit"

temp = str(input("Qual temperatura você escolhe para fazer a conversão? Celsius ou Fahrenheit: "))
if temp == C:
 valor = float(input("Qual temperatura você escolhe: "))
 conversao = (valor*9/5) + 32
 print(f"Sua conversão está feita, aqui está: {conversao:.2f}")
elif temp == F:
    valor = float(input("Digite a temperatura em Fahrenheit: "))
    conversao = (valor - 32) * 5/9
    print(f"Sua conversão está feita, aqui está: {conversao:.2f} °C")
else:
    print("Escolha inválida. Por favor, escolha 'Celsius' ou 'Fahrenheit'.")
```
